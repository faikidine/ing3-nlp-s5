## Avec contexte

![Prompt avec contexte](https://i.ibb.co/4n0g2V3v/Capture-d-cran-2026-09-10-11-41-14.png)

Le promt ici est beaucoup plus mais précis et beaucoup plus contextualisé, ce qui permet d'avoir des réponses beaucoup plus pertinentes et adaptées à la situation (ce prompt lui même je l'ai fait générer pour avoir du contexte "plausible" dans le domaine de la logistique). Mais ici, chaque phrase apporte une information clé :rôle, situation,contraintes, objectif et même le format attendu (5 actions classées) : rend la réponse directement exploitable.

J'imagine qu'on arrive à une réponse satifaisante aussi grâce encore une fois à son "raisonnement",  qui ici est nettement plus long, (*8min*), pourtant la réponse qu'il donne est plus courte que la précédente.
